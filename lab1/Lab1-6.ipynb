4. Как найти самое часто встречаемое значение между двумя массивами? 

````python
from collections import Counter

arr1 = np.random.randint(1, 10, size=10)
arr2 = np.random.randint(1, 10, size=10)

print(arr1, arr2)

combined = np.concatenate((arr1, arr2))
most_common_value, frequency = Counter(combined).most_common(1)[0]
print(f"Самое часто встречаемое значение между двумя массивами: {most_common_value}, встречается {frequency} раз(а).") 
````

8. Создайте случайную матрицу 10х2 которая представляет собой декартовые координаты и преобразуйте их в полярные

````python
cartesian_coords = np.random.random((10, 2)) * 100 

x = cartesian_coords[:, 0]
y = cartesian_coords[:, 1]

r = np.sqrt(x**2 + y**2)
theta = np.arctan2(y, x)

polar_coords = np.column_stack((r, theta))

print(" Декартовы координаты (x, y):\n", cartesian_coords)
print("\n Полярные координаты (r, θ):\n", polar_coords)
````

1. Правда ли, что количество часов затрачиваемых на сон распределено примерно нормально?

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import gaussian_kde

sleep_hours = df['Sleep_Hours']

plt.figure(figsize=(7, 5))

bins = np.arange(3.5, 11, 1) 

# KDE в интервале от 3.5 до 10.5 с помощью seaborn у меня построить не получилось 

# ax = sns.histplot(df['Sleep_Hours'], color = 'blue', bins = bins, kde = True, kde_kws={'bw_adjust': 2.5, 'clip': (3.5, 10.5)}, edgecolor='white', alpha=0.6)
# ax.get_lines()[0].set_color('black')

ax = sns.histplot(sleep_hours, color='blue', bins=bins, edgecolor='white', alpha=0.6)

x = np.linspace(3.5, 10.5, 100)
kde = gaussian_kde(sleep_hours, bw_method=0.45)
ax.plot(x, kde(x) * len(sleep_hours), color='black')

plt.title("Распределение по количеству часов сна")
plt.xlabel("Количество часов сна в сутки")
plt.ylabel("Количество человек")
plt.grid(True)


plt.show()
````

Правда.

Гистограмма с наложенной кривой плотности (KDE) отлично подходит для того, чтобы показать форму распределения. (Для наглядности KDE сгладил) Визуально распределение количества часов сна напоминает нормальное распределение: большинство значений сосредоточено вокруг среднего (7 часов), а частота значений уменьшается по обеим сторонам.

6. Как физическая активность влияет на успеваемость?

````python

# Под успеваемостью я подразумевал Attendance, Previous_Scores и Exam_Score
median_exam_scores = df.groupby('Physical_Activity')['Exam_Score'].median().reset_index()
median_attendance_scores = df.groupby('Physical_Activity')['Attendance'].median().reset_index()
median_previous_scores = df.groupby('Physical_Activity')['Previous_Scores'].median().reset_index()

plt.figure(figsize=(7, 5))
sns.lineplot(x='Physical_Activity', y='Attendance', label='Посещаемсоть занятий', data=median_attendance_scores, marker='o')
sns.lineplot(x='Physical_Activity', y='Previous_Scores', label='Первый экзамен', data=median_previous_scores, marker='o')
sns.lineplot(x='Physical_Activity', y='Exam_Score', label='Конечный экзамен', data=median_exam_scores, marker='o')

plt.title('Влияние физической активности на успеваемость')
plt.xlabel('Количество занятий физической активностью', fontsize = 11)
plt.ylabel('Медиана показателей', fontsize = 11)
plt.legend()
plt.show()

merged_data = pd.merge(median_attendance_scores, median_previous_scores, on='Physical_Activity')
merged_data = pd.merge(merged_data, median_exam_scores, on='Physical_Activity')
merged_data.set_index('Physical_Activity').plot(kind='bar', figsize=(7, 5))

plt.title('Влияние физической активности на успеваемость')
plt.xlabel('Количество занятий физической активностью', fontsize = 11)
plt.ylabel('Медиана показателей', fontsize = 11)
plt.legend(['Посещаемость', 'Первый экзамен', 'Конечный экзамен'])
plt.show()


print(f"Коэффициент корреляции (Посещаемость занятий) - {df['Physical_Activity'].corr(df['Attendance']):.4f}") 
print(f"Коэффициент корреляции (Первый экзамен) - {df['Physical_Activity'].corr(df['Previous_Scores']):.4f}")
print(f"Коэффициент корреляции (Конечный экзамен) - {df['Physical_Activity'].corr(df['Exam_Score']):.4f}") # Задание 2.6
````

Для анализа влияния физической активности на успеваемость были взяты `lineplot`(Линейный график) и `barplot`(Столбчатая диаграмма).

По линейному графику и диаграмме можно сделать вывод о том, что количество занятий физ. активностью не влияет на результаты конечного экзамена. Просчитанный коэффициент корреляции подтверждает наблюдения. (Задание 2.6)

Линейный график демонстрирует отрицательную корреляцию между количеством занятий физ. активностью и медианными показателями результатов за первый экзамен и количества посещаемых занятий. 

Однако, столбчатая диаграмма показывает, что занятия физ. активностью не способствуют значительным изменениям в успеваемости.

Кроме того, во всех случаях корреляция близка к 0 - это указывает на отсутствие линейной зависимости.

7. Постройте HeatMap где ось Х - процент посещаемость, Y - количество занятий с репетитором, а Z - балл за итоговый экзамен*

\* Задание на доп. балл

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

attendance = df["Attendance"]
tutoring_sessions = df["Tutoring_Sessions"]
exam_scores = df["Exam_Score"]

data = pd.DataFrame(list(zip(attendance, tutoring_sessions, exam_scores)), columns=['Attendance', 'Tutoring_Sessions', 'Exam_Score'])

data_aggregated = data.groupby(['Attendance', 'Tutoring_Sessions'], as_index=False).mean()

df_transformed = data_aggregated.pivot(index='Attendance', columns='Tutoring_Sessions', values='Exam_Score')

plt.figure(figsize=(7, 5))

heatmap_plot = sns.heatmap(df_transformed, cmap = 'viridis')

heatmap_plot.set_title("Heatmap \n (Влияние посещаемости и занятий с репетитором на балл за итоговый экзамен)", fontsize=12)
heatmap_plot.set_xlabel("Количество занятий с репетитором", fontsize=10)
heatmap_plot.set_ylabel("Посещаемость в %", fontsize=10)

plt.show()
````
